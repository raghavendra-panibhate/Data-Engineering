```python
# Data structures are a way of organizing and storing data in a computer so that it can be accessed and modified efficiently. 
# Python has several built-in data structures, 

# - set: Sets are unordered collections of unique items.
```

```python
# - List: Lists are one of the most commonly used data structures in Python. 
#         They are ordered, mutable, and can contain duplicate items. 
#         Lists are defined using square brackets [] and can contain items of different data types.

fruits = ["apple", "cherry", "banana", "mango"]
# append: The append() method is used to add an item to the end of a list.
fruits.append("mango")
# remove: The remove() method is used to remove an item from a list.
fruits.remove("banana")
# insert: The insert() method is used to add an item at a specific index in a list.
fruits.insert(0, "grapes")
# sort: The sort() method is used to sort the items in a list.
fruits.sort()
# reverse: The reverse() method is used to reverse the order of the items in a list.
fruits.reverse()

print(fruits)
print(type(fruits))
```

```python
# - Tuple: tuples are ordered, immutable collections of items, and can contain duplicate items.
#          They are defined using parentheses () and can contain items of different data types.

colors = ("red","green","blue","red")

print(colors)
# count: The count() method is used to count the number of occurrences of an item in a tuple.
print(colors.count("red"))
# index: The index() method is used to find the index of the first occurrence of an item in a tuple.
print(colors.index("green"))
print(type(colors))
```

```python
# Dictionaries: Dictionaries are unordered, mutable and no duplicate collections of key-value pairs.
#              They are defined using curly braces {} and can contain items of different data types.

person = {"name":"raghavendra", "age":"28", "designation":"engineer"}
person["city"] = "bengaluru"

# get: The get() method is used to retrieve the value of a specific key in a dictionary.
print(person.get("name"))
print(person["age"])
print(person)
print(type(person))

print("=======================================================")

# keys(): The keys() method is used to retrieve a view object that displays a list of all the keys in a dictionary.
# values(): The values() method is used to retrieve a view object that displays a list of all the values in a dictionary.
# items(): The items() method is used to retrieve a view object that displays a list of all the key-value pairs in a dictionary.
keys = person.keys()
values = person.values()
items = person.items()

keylist = tuple(keys)
print(keylist)
print(type(keylist))

print(keys)
print(type(keys))
print("=======================================================")
print(values)
print(type(values))
print("=======================================================")
print(items)
print(type(items))

#update: The update() method is used to update the value of a specific key in a dictionary.
print(person.update({"age":"29"}))

print(person)
```

```python
# Sets: Sets are unordered(after print it'll order themselves), mutable and no duplicate collections of unique items.
#        They are defined using curly braces {} and do not allow duplicate values.

numbers = {8, 6, 3, 4, 3, 5, 6}

# adding elements to a set
numbers.add(7)
# updating a set with multiple elements
numbers.update([1, 11])
numbers.discard(3)  # remove an element from a set
numbers.remove(4)  # remove an element from a set, raises KeyError if the element is not found
numbers.pop()  # remove and return an arbitrary element from a set
# numbers.clear()  # remove all elements from a set

print(numbers)
print(type(numbers))
```
