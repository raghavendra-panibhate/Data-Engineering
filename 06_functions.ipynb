```python
# arbitrary arguments are used when you do not know how many arguments that will be passed into your function,
# or if you want to pass in a list of values.
def add_all(*numbers):
    total = 0
    for num in numbers:
        total += num
    return total

# numbers = [1,2,3,4,5]
result = add_all(*[1,2,3,4])
print(result)
```

```python
# arbitrary arguments which accepts dictionary(key and values)
def print_details(**details):
    for key,values in details.items():
        print(f"{key}:{values}")
        
print_details(name ='raghavendra',age = 27,designation = 'ASE')
```

```python
# Basic calculator using function

def add(x,y):
    return x+y

def substract(x,y):
    return x-y

def multiply(x,y):
    return x*y

def divide(x,y):
    if y == 0:
        return "Error: division with zero is invalid..."
    return x/y

print("Select Operation: ")
print("1. Addition")
print("2. Substraction")
print("3. Multiplication")
print("4. Division")

choice = input("Enter the the choice 1/2/3/4: ")

if choice in ['1','2','3','4']:
    try:
        num1 = float(input("Enter the Number1: "))
        num2 = float(input("Enter the Number2: "))
        
        if choice == '1':
            print("Result: ",add(num1, num2))
        elif choice == '2':
            print("Result: ",substract(num1, num2))
        elif choice == '3':
            print("Result: ",multiply(num1, num2))
        else:
            print("Result: ",divide(num1, num2))
    except ValueError:
        print("Invalid input ",ValueError)
else:
    print("Invalid input...")
```
