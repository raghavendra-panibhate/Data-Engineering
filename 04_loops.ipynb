```python
# For loop

print("Even Numbers")
for i in range(1, 20):
    if i % 2 == 0:
        print(i)

print("Odd Numbers")
for i in range(1, 20):
    if i % 2 != 0:
        print(i)
```

```python
# While loop

count = 3

while(count > 0):
    print(count)
    count -= 1
```

```python
# break, pass and continue

i = 1

while(i <= 5):
    print(i)
    if i == 3:
        print("Breaking")
        break
    i += 1
```

```python
# Continue

for i in range(1, 5):
    if i == 3:
        print("Skipping..")
        continue
    print(i)
```
