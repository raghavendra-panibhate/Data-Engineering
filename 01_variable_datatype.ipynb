```python
# Variables is the name of the memory location where we store data. 
# In Python, we don't need to declare the type of variable, it is inferred from the value assigned to it.
# rules for naming variables:
#  - Variable names must start with a letter or an underscore
#  - Variable names cannot contain spaces
#  - Variable names are case-sensitive

# Datatypes is the classification of data into different types. Python has several built-in data types, including:
# - int: Integer numbers (e.g., 1, 2, 3)
# - float: Floating-point numbers (e.g., 1.0, 2.5, 3.14)
# - str: Strings (e.g., "Hello", "World")
# - bool: Boolean values (True or False)

name = "Raghavendra"
age = 28
height = 5.9
is_student = True

print(f"My name is {name} and I am {age} years old.")
print(f"Height: {height}")
print(f"Is a student: {is_student}")

print(type(name))
print(type(age))
print(type(height))
print(type(is_student))
```
