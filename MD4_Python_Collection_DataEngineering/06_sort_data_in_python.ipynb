- sorted function is used to the sort data in asc form
- sorted(data/iterable, key)

```python
people = [
    {"name":"bheema", "age":28},
    {"name":"raghu", "age":27},
    {"name":"sharat", "age":31},
    {"name":"ullas", "age":29}
]

sorted_values = sorted(people, key = lambda x : x['age'], reverse=False)
print(sorted_values)

names = ['raghavendra', 'bheema', 'ullas']
sort_lenth = sorted(names, key=len, reverse=True)
print(sort_lenth)

numbers ={1,3,6,10,9,2,8,3}
sort = sorted(numbers)
print(sort)
```
