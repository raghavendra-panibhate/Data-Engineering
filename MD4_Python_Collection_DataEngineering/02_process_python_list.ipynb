========== Processing Python List ==========

```python
data = [1,2,3,4,5]

square = [x**2 for x in data]
print(square)

even = [x for x in data if x % 2 == 0]
print(even)

odd = [x for x in data if x % 2 != 0]
print(odd)

total = sum(data)
print(total)

avarage = sum(data)/len(data)
print(avarage)
```
