=========== Read data from CSV file ===========

```python
import csv as c

with open("data/data.csv", 'r') as file:
    reader = c.reader(file)
    print(reader)
    data = [i for i in reader]

print(data)
```

```python
import csv as csv

with open("data/data.csv", 'r') as file:
    reader = csv.DictReader(file) # used to read the data into dictionary
    print(reader)
    for row in reader:
        print(row)
    data = [i for i in reader]
    print(data)
```

=========== Python Collection ===========

```python
from collections import namedtuple, defaultdict, Counter

# named tuple is used for access fields by using there name instead of remembering the index
emp = namedtuple('employee',['name', 'age', 'salary'])
user = emp(name='raghu', age=27, salary=100000)
user1 = emp(name='Bheema', age=28, salary=200000)
print(user)
print(user.name)
print(user1)
print(user1.name)
print(user1[0])
```

```python
# defaultdict -- Normal dictionary give error when you don't have key in there, but defaultdict will create the key.

from collections import namedtuple, defaultdict, Counter

d = defaultdict(list)
d['IT'].append('Raghu')
d['IT'].append('Bheema')
d['Govt'].append('Ullas')
d['Govt'].append('Mahesh')
print(d)
```

```python
from collections import namedtuple, defaultdict, Counter
# The slow, boring way
snacks = ["apple", "cookie", "apple", "juice", "cookie", "apple"]

tally = {}
for item in snacks:
    if item not in tally:
        tally[item] = 0
    tally[item] += 1

print(tally)

# Counter -- resolve the above loop

inventory = Counter(snacks)
print(inventory)
```
