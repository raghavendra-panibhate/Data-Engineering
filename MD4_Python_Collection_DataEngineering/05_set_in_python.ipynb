set() function in python
- removing the duplicate
- performing set operations(union, intersection etc.)

```python
name = ['raghu','bheema','ullas','vinay','raghu']

unique_values = list(set(name))
print(unique_values)
```

```python
name = {'raghu','ullas','bheema'}
name1 = {'raghu','bheema'}

non_common = name - name1
print(non_common)
```

```python
# find the unique values from the dict
employee = [
    {"name":"raghu", "salary":10000},
    {"name":"bheema", "salary":20000},
    {"name":"ullas", "salary":10000},
]

unique_values = set(map(lambda x : x['salary'], employee)) # set function will autometically remove the duplicate values
print(unique_values)
```
