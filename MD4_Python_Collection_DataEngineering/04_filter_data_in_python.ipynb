```python
data = [100,200,300,400]

result = list(filter(lambda x :x>150, data))
print(result)
```

```python
# clean the data

data =[100,0,5,-2,200, None]

cleaned = list(filter(lambda x: x is not None and x>0, data))
print(cleaned)
```

```python
# filter the data from the dictionary list

employee = [
    {"name":"raghu", "salary":10000},
    {"name":"bheema", "salary":20000},
    {"name":"ullas", "salary":30000},
]

high_salary = list(filter(lambda emp: emp["salary"]>10000, employee))
print(high_salary)

salary = list(filter(lambda x : x["salary"] == 20000, employee))
print(salary)
```
