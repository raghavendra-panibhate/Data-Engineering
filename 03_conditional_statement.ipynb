```python
# Conditional statements execute different code branches depending on whether a boolean expression evaluates to True or False.

score = int(input("Enter your score: "))

print(f"Score is: {score}")

if score <= 100:
    if score >= 90:
        print("Grade: A")
    elif score >= 70 and score <= 90:
        print("Grade: B")
    elif score >= 40 and score <= 70:
        print("Grade: C")
    else:
        print("Grade: D")
else:
    print("Invalid Marks.")
```
